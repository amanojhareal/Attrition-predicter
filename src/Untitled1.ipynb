```python
# Find categorical columns
categorical_cols = df.select_dtypes(include='str').columns.tolist()

# Separate categorical columns based on unique values
one_hot_cols = []
binary_cols = []

for col in categorical_cols:
    unique_count = df[col].nunique()

    if unique_count <= 2:
        one_hot_cols.append(col)
    else:
        binary_cols.append(col)

print("One-Hot columns:")
print(one_hot_cols)

print("\nBinary columns:")
print(binary_cols)
```

```python
one_hot_cols =[]
binary_cols=[]
map_cols =[]

for col in categorical_cols:
    unique_count = df[col].nunique()
    
    if unique_count == 2:
        map_cols.append(col)
        
    
    elif unique_count < 5:
        one_hot_cols.append(col)
        
    else:
        binary_cols.append(col)

print(map_cols)
print(one_hot_cols)
print(binary_cols)
```
